```python
import pandas as pd

df = pd.DataFrame({
    "height":[160, 162, 165, 167, 169, 170, 172, 174, 175, 176,
              168, 169, 171, 173, 166, 164, 167, 170, 172, 180]
})

print("Mean:", df["height"].mean())
print("Median:", df["height"].median())
print("Mode:", df["height"].mode()[0])
print("Min:", df["height"].min())
print("Max:", df["height"].max())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "height":[158,160,162,164,166,168,170,172,174,176,
              165,167,169,171,173,175,168,170,172,180]
})

df.boxplot(column="height")
plt.title("Box Plot of Height")
plt.ylabel("Height")
plt.show()
```

```python
# Detecting outliers using IQR
import pandas as pd

df = pd.DataFrame({
    "values":[5, 7, 8, 9, 10, 150]
})

Q1 = df["values"].quantile(0.25)
Q3 = df["values"].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df["values"] < Q1 - 1.5*IQR) | 
              (df["values"] > Q3 + 1.5*IQR)]

print('Outliers', outliers['values'].values)
```

```python
# calculating percentile
import pandas as pd

df = pd.DataFrame({
    "values":[6, 8, 9, 11, 13, 120]
})

print("25th percentile:", df["values"].quantile(0.25))
print("50th percentile:", df["values"].quantile(0.50))
print("75th percentile:", df["values"].quantile(0.75))
```

```python
import numpy as np

scores = [75, 85, 95]
weights = [0.2, 0.3, 0.5]

weighted_mean = np.average(scores, weights=weights)

print(weighted_mean)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "study_hours":[1,3,5,7],
    "exam_score":[45,60,70,85]
}

df = pd.DataFrame(data)

print(df.corr())
```

```python
# Time series data plot
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Sales":[180,210,260,230,270]
}

df = pd.DataFrame(data)

plt.plot(df["Month"], df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
